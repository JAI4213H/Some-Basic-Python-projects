```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = Image.open("lol.jpg").convert("L")
```

```python
a = np.array(image, dtype=float)
```

```python
a.size
```

```python
U, S, VT = np.linalg.svd(a,full_matrices=False)
```

```python
print(U.shape)
print(S.shape)
print(VT.shape)
```

```python
first_8 = []
for i in range(10,50,5):
  summ = np.zeros_like(a,dtype=float)
  for j in range(i):
    summ += S[j]*np.outer(U[:,j],VT[j,:])
  first_8.append(summ)
```

```python
fig, axes = plt.subplots(2,4,figsize=(15,10))
for i in range(8):

  col = i % 4
  row = i//4
  axes[row,col].imshow(first_8[i],cmap="grey")
  axes[row,col].axis("off")
  axes[row,col].set_title(f"rank = {i+10}")
plt.tight_layout()
```
